```python
# imports

import os
import requests
from pathlib import Path
from dotenv import load_dotenv
```

```python
# load environment variables

env_path = Path("/Users/oz/Documents/personal/education/dividend_stock_portfolio/.env")

print(env_path.exists())

load_dotenv(dotenv_path=env_path)
```

```python
# get the fmp api key

api_key = os.getenv("FMP_API_KEY")

if not api_key:
    raise ValueError("Missing FMP_API_KEY in .env file")

print("FMP API key loaded successfully")
```

```python
# set ticker
symbol = "AAPL"

# build fmp url
url = "https://financialmodelingprep.com/stable/cash-flow-statement"

params = {
    "symbol": symbol,
    "apikey": api_key,
}
```

```python
# api request
response = requests.get(url, params=params, timeout=30)

# hide api key
assert api_key is not None
safe_url = response.url.replace(api_key, "REDACTED") if response.url else "URL not available"

print("Final URL:")
print(safe_url)

print("\nStatus code:")
print(response.status_code)

print("\nResponse Preview:")
print(response.text[:1000])
```

```python
# convert reponse to python json object
data = response.json()
```

```python
print(type(data))

print(len(data))

data[0]
```

```python
record_latest = data[0]
matching_records_fcf = []

for key in record_latest.keys():
    if "free" in key.lower():
        matching_records_fcf.append(key)

matching_records_fcf
```

```python
matching_records_dividendspaid = []

for key in record_latest.keys():
    if "div" in key.lower():
        matching_records_dividendspaid.append(key)
        print(key,":", record_latest[key])

matching_records_dividendspaid
```
