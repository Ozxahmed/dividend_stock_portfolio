```python
from src.config import load_env_variable
import requests
import pandas as pd
```

```python
# load fmp api key
api_key = load_env_variable("FMP_API_KEY")

# set ticker
symbol = "AAPL"
```

```python
# api endpoint url
endpoint_url = "https://financialmodelingprep.com/stable/income-statement"

# api params
params = {
    "symbol": symbol,
    "apikey": api_key,
}

# api call
response = requests.get(url=endpoint_url, params=params, timeout=30)

# check status code
response.status_code # 200
```

```python
# mask api key
safe_url = response.url.replace(api_key, "REDACTED")

print(safe_url)
print(response.status_code)
print(response.text[:1000])
```

```python
# convert response to python json object
data = response.json()

# check type
print(type(data))
```

```python
def search_df_columns(df:pd.DataFrame, name:str | list[str]) -> list:
    matching_records = []

    if isinstance(name,str):
        name = [name]

    for search_term in name:
        for column in df.columns:
            if search_term in column:
                matching_records.append(column)

    return matching_records
```
